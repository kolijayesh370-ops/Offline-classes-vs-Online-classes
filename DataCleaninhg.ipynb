```python
# ==========================================
# NOTEBOOK 1: DATA CLEANING & STANDARDIZATION
# ==========================================
import os
import io
import pandas as pd
from google.colab import files

# Create data directory
os.makedirs("data", exist_ok=True)

# 1. Ingest raw survey responses
print("Upload your raw survey file (.xlsx or .csv):")
uploaded = files.upload()
raw_filename = list(uploaded.keys())[0]

if raw_filename.lower().endswith('.csv'):
    df_raw = pd.read_csv(io.BytesIO(uploaded[raw_filename]))
else:
    df_raw = pd.read_excel(io.BytesIO(uploaded[raw_filename]))

# Preserve raw copy for Chapter 8.1 Evidences
df_raw.to_excel("data/raw_responses.xlsx", index=False)
print(f"Loaded raw records: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns.")

# 2. Data Cleaning
df_clean = df_raw.copy()
df_clean.columns = [col.strip() for col in df_clean.columns]

column_mapping = {
    'Timestamp': 'timestamp',
    'Email address': 'email',
    'Respondent Type': 'respondent_type',
    'Age': 'age_group',
    'Gender': 'gender',
    'Education Level': 'education_level',
    'Which mode of learning do you think is more effective?': 'overall_effectiveness',
    'Which mode helps students understand concepts better?': 'concept_understanding',
    "Which mode improves students' academic performance more?": 'academic_performance',
    'Which mode provides better interaction between teachers and students?': 'teacher_student_interaction',
    'How would you rate student engagement during online classes?': 'engagement_online',
    'How would you rate student engagement during offline classes?': 'engagement_offline',
    'What is the biggest challenge of online classes?': 'challenge_online',
    'What is the biggest challenge of offline classes?': 'challenge_offline',
    'Which mode helps students stay more disciplined?': 'discipline_mode',
    'Do you think online classes should continue along with offline classes?': 'continue_online',
    'Which learning model do you prefer in the future?': 'future_preference',
    'Please share any suggestions for improving online and offline learning': 'suggestions'
}
df_clean.rename(columns=column_mapping, inplace=True)

# Standardize non-standard response choices
df_clean['challenge_online'] = df_clean['challenge_online'].replace({
    'Internet issues, concentration ,Technical problems ,Home distractions': 'Multiple Challenges'
})
df_clean['challenge_offline'] = df_clean['challenge_offline'].replace({
    'Nothing': 'No Major Challenge'
})

# Impute empty text comments
df_clean['suggestions'] = df_clean['suggestions'].fillna('No suggestion provided')

# Standardize comparative categories
standardize_cols = ['concept_understanding', 'academic_performance', 'discipline_mode', 'teacher_student_interaction']
for col in standardize_cols:
    df_clean[col] = df_clean[col].replace({'Both equally': 'Both Equally', 'Both': 'Both Equally'})

# 3. Export cleaned dataset and download to computer
output_path = "data/cleaned_survey_data.csv"
df_clean.to_csv(output_path, index=False)
print("Data cleaned successfully! Downloading file...")
files.download(output_path)
```
