```python
# ==========================================
# EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================
import io
import pandas as pd
from google.colab import files

print("Upload 'cleaned_survey_data.csv':")
uploaded = files.upload()
csv_name = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[csv_name]))

print(f"Data ready for analysis. Total Respondents (N): {len(df)}")

print("=" * 60)
print(" DEMOGRAPHIC PROFILE")
print("=" * 60)
for field in ['respondent_type', 'gender', 'age_group', 'education_level']:
    counts = df[field].value_counts()
    pcts = (df[field].value_counts(normalize=True) * 100).round(1)
    table = pd.DataFrame({'Count': counts, 'Percentage (%)': pcts})
    print(f"\n[{field.upper()}]\n{table}")

print("\n" + "=" * 60)
print("CORE COMPARATIVE METRICS")
print("=" * 60)
for metric in ['overall_effectiveness', 'concept_understanding', 'academic_performance',
               'teacher_student_interaction', 'discipline_mode', 'future_preference']:
    counts = df[metric].value_counts()
    pcts = (df[metric].value_counts(normalize=True) * 100).round(1)
    table = pd.DataFrame({'Count': counts, 'Percentage (%)': pcts})
    print(f"\n[{metric.upper()}]\n{table}")

print("\n" + "=" * 60)
print("ENGAGEMENT SCORES SUMMARY")
print("=" * 60)
engagement_comparison = pd.DataFrame({
    'Online Engagement': df['engagement_online'].describe(),
    'Offline Engagement': df['engagement_offline'].describe()
}).round(2)
print(engagement_comparison)
```
